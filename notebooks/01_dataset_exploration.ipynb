Comprehensive exploration of the Diabetic Retinopathy retinal dataset (APTOS 2019 / EyePACS benchmark standard) with class distributions, statistical analysis, and demographic distributions.


## Environment & Setup
Import required libraries for tabular and image visualization.


```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

sns.set_theme(style='whitegrid')
print('Environment initialized successfully.')
```

## Inspect Class Labels & Metadata
Examine class distribution and severity grades across the benchmark dataset.


```python
classes = ['No DR', 'Mild', 'Moderate', 'Severe', 'Proliferative DR']
counts = [1805, 370, 999, 193, 295]
df_stats = pd.DataFrame({'Class ID': range(5), 'Severity': classes, 'Patient Count': counts})
df_stats['Percentage'] = (df_stats['Patient Count'] / df_stats['Patient Count'].sum() * 100).round(2)
df_stats
```

## Class Imbalance Visualization
Class 0 (No DR) represents ~49% of the dataset, highlighting the severe class imbalance in real-world clinical screening populations.


```python
plt.figure(figsize=(10, 5))
bars = plt.bar(df_stats['Severity'], df_stats['Patient Count'], color=['#2b5c8f', '#4682b4', '#e67e22', '#d35400', '#c0392b'])
plt.title('Diabetic Retinopathy Severity Class Distribution (APTOS 2019)')
plt.xlabel('Diagnosis Grade')
plt.ylabel('Number of Retinal Fundus Images')
for bar in bars:
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2.0, yval + 20, int(yval), ha='center', va='bottom', fontweight='bold')
plt.tight_layout()
plt.show()
```
