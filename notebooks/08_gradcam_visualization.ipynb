## Grad-CAM++ Pipeline
Initialize GradCAMExplainer targeting EfficientNet features block.


```python
from ml.explainability.gradcam import GradCAMExplainer
from ml.models.cnn_model import DRClassifier
import torch

model = DRClassifier(num_classes=5, pretrained=False)
explainer = GradCAMExplainer(model=model, target_layer_name='backbone.features.8', device='cpu')
print('Grad-CAM Explainer initialized on target layer:', explainer.target_layer_name)
```
