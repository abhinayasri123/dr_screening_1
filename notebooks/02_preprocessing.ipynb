Verify horizontal flips, restricted rotations (<=15 deg to maintain anatomic fidelity), contrast/brightness adjustments, and Gaussian blur.


```python
preprocessor = RetinalImagePreprocessor(image_size=300)
train_transforms = preprocessor.get_train_transforms()
val_transforms = preprocessor.get_val_transforms()
print('Training Transformations:\n', train_transforms)
print('\nValidation Transformations:\n', val_transforms)
```
