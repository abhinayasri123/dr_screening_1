Late fusion model dynamically balancing convolutional retinal fundus representations with tabular clinical health indicators using an Attention Gate.


## Multimodal Fusion Architecture
Pass both image and EHR modalities through the attention fusion gating mechanism.


```python
import torch
from ml.models.multimodal_model import MultimodalDRClassifier

model = MultimodalDRClassifier(num_classes=5, ehr_input_dim=17, mode='multimodal')
img = torch.randn(2, 3, 300, 300)
ehr = torch.randn(2, 17)

out = model(image=img, ehr=ehr)
print('Multimodal Logits:', out['logits'].shape)
print('Image Feature Dimension:', out['image_features'].shape)
print('EHR Feature Dimension:', out['ehr_features'].shape)
```

## Ablation Modes: Image-Only vs EHR-Only vs Multimodal
Showcase support for all 3 ablation modes within a single unified model.


```python
model.set_mode('image_only')
res_img = model(image=img)
model.set_mode('ehr_only')
res_ehr = model(ehr=ehr)
print('Image-only forward pass:', res_img['logits'].shape)
print('EHR-only forward pass:', res_ehr['logits'].shape)
```
