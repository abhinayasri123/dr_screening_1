EfficientNet-B3 deep convolutional network training, feature extractor verification, and backbone fine-tuning strategy.


## Model Instantiation
Load EfficientNet-B3 with replaced 5-class linear classifier and dropout regularizers.


```python
import torch
from ml.models.cnn_model import DRClassifier

model = DRClassifier(num_classes=5, pretrained=False)
print('Backbone:', model.backbone.__class__.__name__)
print('Classification Head:', model.backbone.classifier)
```

## Feature Extraction Verification
Confirm feature map pooling outputs a 1536-dimensional latent visual vector.


```python
dummy_input = torch.randn(2, 3, 300, 300)
feature_extractor = model.get_feature_extractor()
with torch.no_grad():
    features = feature_extractor(dummy_input)
print('Extracted visual representation vector shape:', features.shape)
```

## Warmup & Fine-Tuning Strategy
Demonstrates freezing convolutional feature extraction layers during initial warmup epochs.


```python
model.freeze_backbone()
frozen_params = sum(p.requires_grad for p in model.parameters())
model.unfreeze_backbone()
unfrozen_params = sum(p.requires_grad for p in model.parameters())
print(f'Trainable parameters when frozen: {frozen_params}')
print(f'Trainable parameters when unfrozen: {unfrozen_params}')
```
