## Multi-Layer Perceptron (MLP) Encoder
Verify the 3-layer MLP encoder reduces sparse clinical tabular records to a 64-dimensional dense representation.


```python
import torch
from ml.models.ehr_encoder import EHREncoder

encoder = EHREncoder(input_dim=17, hidden_dims=[128, 64], output_dim=64)
x = torch.randn(4, 17)
out = encoder(x)
print('Encoded clinical representation shape:', out.shape)
```
